```python
import glob
import mdtraj as md
import nglview as nv

folder = "fol-clustering-shuyu-nosingle/100ns"           # output folder of your cluster_prune.py run
files = sorted(glob.glob(f"{folder}/representative_*.pdb"))
print(files)

structs = [md.load(f) for f in files]

# superpose all on the benzene ring so differences elsewhere are easy to see
ring = structs[0].topology.select("name C4 C5 C6 C7 C8 C9")
print("ring atoms used for alignment:", ring)
for s in structs[1:]:
    s.superpose(structs[0], atom_indices=ring)

colors = ["blue", "orange", "green", "red", "purple"]
view = nv.NGLWidget()
for k, s in enumerate(structs):
    c = view.add_trajectory(s)
    c.clear_representations()
    c.add_ball_and_stick(color=colors[k % len(colors)])
view
```
